# Conclusion

The Exploratory Data Analysis of the Netflix dataset revealed several useful insights.

- The dataset contains more Movies than TV Shows.
- Movies account for approximately 69.62% of the content, while TV Shows account for approximately 30.38%.
- Missing values were mainly observed in Director, Cast, and Country columns.
- No duplicate records were found.
- Release-year analysis showed different trends in Netflix content over time.
- Genre and rating analysis helped identify the major types of content available on Netflix.
- Movie duration and TV Show season analysis provided additional insights into content characteristics.
- The analysis also examined older release years to identify unusual patterns.
- Visualizations were used to identify and communicate important trends and patterns.

Overall, the EDA successfully identified important trends, patterns, data-quality issues, and characteristics of Netflix content.
